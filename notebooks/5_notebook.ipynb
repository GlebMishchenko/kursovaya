```python
def grab_col_names(dataframe, cat_th=10, car_th=20):

    cat_cols = [col for col in dataframe.columns if dataframe[col].dtypes == "O"]
    num_but_cat = [col for col in dataframe.columns if dataframe[col].nunique() < cat_th and dataframe[col].dtypes != "O"]
    cat_but_cat = [col for col in dataframe.columns if dataframe[col].nunique() > car_th and dataframe[col].dtypes == "O"]
    cat_cols = cat_cols + num_but_cat
    cat_cols = [col for col in cat_cols if col not in cat_but_cat]

    num_cols = [col for col in dataframe.columns if dataframe[col].dtypes != "O"]
    num_cols = [col for col in num_cols if col not in num_but_cat]

    print(f"Observations: {dataframe.shape[0]}")
    print(f"Variables: {dataframe.shape[1]}")
    print(f'cat_cols: {len(cat_cols)}')
    print(f'num_cols: {len(num_cols)}')
    print(f'cat_but_cat: {len(cat_but_cat)}')

    return cat_cols, num_cols, cat_but_cat


cat_cols, num_cols, cat_but_cat, = grab_col_names(train_df)
cat_cols, num_cols, cat_but_cat
```

```python
def cat_summary(dataframe, col_name, plot=False):
    print(pd.DataFrame({col_name: dataframe[col_name].value_counts(),
                        "Ratio": 100 * dataframe[col_name].value_counts() / len(dataframe)}))

    if plot:
        fig, axs = plt.subplots(1, 2, figsize=(8, 6))
        plt.subplot(1, 2, 1)
        sns.countplot(x=dataframe[col_name], data=dataframe)
        plt.title("Frequency of " + col_name)
        plt.xticks(rotation=90)

        plt.subplot(1, 2, 2)
        values = dataframe[col_name].value_counts()
        plt.pie(x=values, labels=values.index, autopct=lambda p: '{:.2f}% ({:.0f})'.format(p, p/100 * sum(values)))
        plt.title("Frequency of " + col_name)
        plt.legend(labels=['{} - {:.2f}%'.format(index, value/sum(values)*100) for index, value in zip(values.index, values)],
                   loc='upper center', bbox_to_anchor=(0.5, -0.2), fancybox=True, shadow=True, ncol=1)
        plt.show(block=True)

for col in cat_cols:
    cat_summary(train_df, col, True)
```

```python
def my_histplot(df, col, ax):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(f'Histogram Plot of {col}')
def my_distplot(df, col, ax):
    sns.distplot(df[col], ax=ax)
    ax.set_title(f'Distribution Plot of {col}')
def my_kdeplot(df, col, ax):
    sns.kdeplot(df[col], ax=ax, fill=True)
    ax.set_title(f'KDE Plot of {col}')

def my_scatterplot(df, col, ax):
    sns.scatterplot(df[col], ax=ax)
    ax.set_title(f'Scatter Plot of {col}')
def my_lineplot(df, col, ax):
    sns.lineplot(df[col], ax=ax)
    ax.set_title(f'Line Plot of {col}')

def my_pie_chart(df, col, ax):
    labels = df[col].value_counts()
    ax.pie(labels, labels=labels.index, autopct='%1.1f%%')
    ax.set_title(f'Pie Chart of {col}')
def my_countplot(df, col, ax):
    sns.countplot(x=df[col], ax=ax)
    ax.set_title(f'Count Plot of {col}')
    ax.set_xticklabels(ax.get_xticklabels(), rotation=90)
def my_boxplot(df, col, ax):
    sns.boxplot(y=df[col], ax=ax)
def my_violinplot(df, col, ax):
    sns.violinplot(y=df[col], ax=ax)

def my_heatmap(df, size):
    if size: plt.figure(figsize=size)
    sns.heatmap(df.corr(), annot=True, fmt=".1f", cmap='Blues', annot_kws={"size": 12})
    plt.title('Correlation Heatmap')
    plt.show()

def my_vsplot(df, normal_col, label_col):
    plt.figure(figsize=(10, 6), dpi=80)
    plt.bar(list(dict(df[normal_col].value_counts()).keys()), dict(df[normal_col].value_counts()).values(), color='r')
    plt.bar(list(dict(df[normal_col][df[label_col] == 1].value_counts()).keys()), dict(df[normal_col][df[label_col] == 1].value_counts()).values(), color='b')

    plt.xlabel(normal_col)
    plt.ylabel('Count')
    plt.legend(['All', label_col])

def plot_charts_grid_single_feature(df, plot_func, size=(12, 4), n_col=1):
    if len(df.columns) == 0:
        return
    n_rows = (len(df.columns) + n_col-1) // n_col
    fig, axes = plt.subplots(n_rows, n_col, figsize=(size[0]*n_col, size[1]*n_rows))
    if len(df.columns) == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    for i, label in enumerate(df.columns):
        plot_func(df, label, axes[i])
        axes[i].set_xlabel(label)

    for j in range(i+1, n_rows*n_col):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE
from lightautoml.automl.presets.tabular_presets import TabularAutoML
from lightautoml.tasks import Task
from collections import Counter
import seaborn as sns
import matplotlib.pyplot as plt

def prepare_data(train_df, test_df):
    """Подготовка данных с гарантированным наличием всех классов"""
    # Объединяем редкие классы
    rare_classes = ['NetBIOS', 'LDAP']
    train_df['Label'] = train_df['Label'].replace({k: 'RARE' for k in rare_classes})
    test_df['Label'] = test_df['Label'].replace({k: 'RARE' for k in rare_classes})

    # Фиксируем кодировщик на train данных
    le = LabelEncoder()
    le.fit(train_df['Label'])

    # Кодируем метки
    train_df['Label_encoded'] = le.transform(train_df['Label'])
    test_df['Label_encoded'] = le.transform(test_df['Label'])

    return train_df, test_df, le

def analyze_classes(df, le):
    """Анализ распределения классов"""
    print("\nДетальное распределение:")
    for cls_code, cls_name in enumerate(le.classes_):
        count = sum(df['Label_encoded'] == cls_code)
        print(f"{cls_name} (код {cls_code}): {count} примеров")

def safe_smote_balance(train_df, le, min_samples=50):
    """Безопасная балансировка с проверкой условий"""
    # Выделяем фичи и таргет
    X = train_df.drop(['Label', 'Label_encoded'], axis=1)
    y = train_df['Label_encoded']

    # Анализируем классы перед балансировкой
    analyze_classes(train_df, le)

    # Определяем стратегию увеличения только для минорных классов
    strategy = {}
    for cls in np.unique(y):
        count = sum(y == cls)
        if count < min_samples:
            continue  # Пропускаем слишком малочисленные классы
        target_size = min(count * 3, 50000)  # Увеличиваем не более чем в 3 раза
        strategy[cls] = target_size

    if not strategy:
        print("\nНет классов для балансировки!")
        return train_df

    print("\nСтратегия балансировки:", {le.classes_[k]:v for k,v in strategy.items()})

    # Применяем SMOTE
    try:
        smote = SMOTE(sampling_strategy=strategy, k_neighbors=2)
        X_res, y_res = smote.fit_resample(X, y)
        return pd.concat([pd.DataFrame(X_res), pd.DataFrame(y_res, columns=['Label_encoded'])], axis=1)
    except ValueError as e:
        print(f"\nОшибка балансировки: {e}")
        print("Возвращаем оригинальные данные")
        return train_df

def train_and_validate(train_df, test_df, le):
    """Обучение модели с контролем качества и возвратом предсказаний"""
    # Инициализация модели
    task = Task('multiclass', metric='f1_weighted')

    automl = TabularAutoML(
        task=task,
        timeout=3660,
        cpu_limit=4,
        general_params={
            'use_algos': [['lgb', 'cb']],
            'nn_params': {'use_qt': False}
        }
    )

    # Обучение
    roles = {'target': 'Label_encoded'}
    automl.fit_predict(train_df, roles=roles)

    # Проверка наличия всех классов в тесте
    test_classes = set(test_df['Label_encoded'].unique())
    train_classes = set(train_df['Label_encoded'].unique())
    missing_classes = test_classes - train_classes

    if missing_classes:
        print(f"\nПредупреждение: В тесте есть классы, отсутствующие в трейне: {missing_classes}")
        test_df = test_df[~test_df['Label_encoded'].isin(missing_classes)]

    # Предсказание и оценка
    if not test_df.empty:
        test_pred = automl.predict(test_df)
        y_pred = test_pred.data.argmax(axis=1)  # Получаем предсказанные классы
        y_true = test_df['Label_encoded'].values  # Истинные метки

        print("\nОтчёт о классификации:")
        print(classification_report(
            y_true,
            y_pred,
            target_names=le.classes_,
            digits=4
        ))

        return automl, y_true, y_pred
    else:
        print("\nТестовые данные не содержат известных классов!")
        return automl, None, None

# Главная функция
def main_pipeline(train_df, test_df):
    """Полный пайплайн обработки с возвратом предсказаний"""
    # 1. Подготовка данных
    train_df, test_df, le = prepare_data(train_df.copy(), test_df.copy())

    # 2. Балансировка
    balanced_train = safe_smote_balance(train_df, le)

    # 3. Обучение и валидация
    model, y_true, y_pred = train_and_validate(balanced_train, test_df, le)

    return model, le, y_true, y_pred

# Запуск
if __name__ == "__main__":


    # Выполнение пайплайна
    automl_model, label_encoder, y_true, y_pred = main_pipeline(train_df, test_df)

    # Дополнительный анализ
    if y_pred is not None:
        print("\nПримеры предсказаний:")
        print("True:", y_true[:10])
        print("Pred:", y_pred[:10])

        print("\nРаспределение предсказаний:")
        print(pd.Series(label_encoder.inverse_transform(y_pred)).value_counts())

        print("\nРаспределение истинных меток:")
        print(pd.Series(label_encoder.inverse_transform(y_true)).value_counts())
```

```python
def plot_confusion_matrix(y_true, y_pred, label_encoder, normalize=None):
    """
    Построение тепловой карты матрицы ошибок

    Параметры:
    y_true - истинные метки классов (массив numpy)
    y_pred - предсказанные метки классов (массив numpy)
    label_encoder - обученный LabelEncoder с классами
    normalize - способ нормализации:
                None - абсолютные значения,
                'true' - по строкам (recall),
                'pred' - по столбцам (precision),
                'all' - по всей матрице
    """
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.metrics import confusion_matrix
    import numpy as np

    # Вычисляем матрицу ошибок
    cm = confusion_matrix(y_true, y_pred)

    # Нормализация при необходимости
    if normalize == 'true':
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        fmt = '.2f'
        vmin, vmax = 0, 1
    elif normalize == 'pred':
        cm = cm.astype('float') / cm.sum(axis=0)[np.newaxis, :]
        fmt = '.2f'
        vmin, vmax = 0, 1
    elif normalize == 'all':
        cm = cm.astype('float') / cm.sum()
        fmt = '.2f'
        vmin, vmax = 0, 1
    else:
        fmt = 'd'
        vmin, vmax = None, None

    # Настройка визуализации
    plt.figure(figsize=(12, 10))
    sns.set(font_scale=1.2)

    # Построение тепловой карты
    ax = sns.heatmap(
        cm,
        annot=True,
        fmt=fmt,
        cmap='Blues',
        cbar=True,
        square=True,
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_,
        vmin=vmin,
        vmax=vmax
    )

    # Настройка подписей
    ax.set_xlabel('Предсказанные классы', fontsize=14)
    ax.set_ylabel('Истинные классы', fontsize=14)

    title = 'Матрица ошибок'
    if normalize == 'true':
        title += ' (нормализована по строкам - Recall)'
    elif normalize == 'pred':
        title += ' (нормализована по столбцам - Precision)'
    elif normalize == 'all':
        title += ' (нормализована)'

    ax.set_title(title, fontsize=16, pad=20)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

```python
def prepare_data(train_df, test_df):
    """Подготовка данных с одинаковым набором признаков"""
    # Удаляем лишние колонки если они есть
    for df in [train_df, test_df]:
        if 'is_attack' in df.columns:
            df.drop('is_attack', axis=1, inplace=True, errors='ignore')

    # Объединяем редкие классы
    rare_classes = ['NetBIOS', 'LDAP']
    train_df['Label'] = train_df['Label'].replace({k: 'RARE' for k in rare_classes})
    test_df['Label'] = test_df['Label'].replace({k: 'RARE' for k in rare_classes})

    # Фиксируем кодировщик
    le = LabelEncoder()
    le.fit(train_df['Label'])

    # Кодируем метки
    train_df['Label_encoded'] = le.transform(train_df['Label'])
    test_df['Label_encoded'] = le.transform(test_df['Label'])

    return train_df, test_df, le

def balance_data(train_df, le):
    """Балансировка с гарантией одинаковых признаков"""
    features = train_df.drop(['Label', 'Label_encoded'], axis=1).columns
    X = train_df[features]  # Явный выбор колонок
    y = train_df['Label_encoded']

    class_counts = Counter(y)
    strategy = {
        cls: min(count * 2, 50000)
        for cls, count in class_counts.items()
        if count < 30000
    }

    if strategy:
        smote = SMOTE(sampling_strategy=strategy, k_neighbors=3)
        X, y = smote.fit_resample(X, y)

    return X, y

def train_boosting(X_train, y_train, model_type='xgb'):
    """Обучение с фиксированными feature_names"""
    model = XGBClassifier(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.1,
        objective='multi:softmax',
        random_state=42
    ) if model_type == 'xgb' else CatBoostClassifier(
        iterations=500,
        depth=6,
        learning_rate=0.1,
        loss_function='MultiClass',
        verbose=0,
        random_state=42
    ) if model_type == 'catboost' else LGBMClassifier(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.1,
        objective='multiclass',
        random_state=42
    )

    model.fit(X_train, y_train)
    return model, X_train.columns.tolist()  # Возвращаем имена фичей

def boosting_pipeline(train_df, test_df, model_type='xgb'):
    # 1. Подготовка
    train_df, test_df, le = prepare_data(train_df, test_df)

    # 2. Балансировка
    X_train, y_train = balance_data(train_df, le)

    # 3. Обучение
    model, feature_names = train_boosting(X_train, y_train, model_type)

    # 4. Предсказание с проверкой фичей
    X_test = test_df[feature_names]  # Только нужные колонки
    y_test = test_df['Label_encoded']
    y_pred = model.predict(X_test)

    print(classification_report(y_test, y_pred, target_names=le.classes_, digits=4))
    return model, le

# Запуск
model, label_encoder = boosting_pipeline(train_df.copy(), test_df.copy(), model_type='xgb')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np
import pandas as pd

def plot_confusion_matrix_and_heatmap(model, test_df, label_encoder, feature_names):
    """
    Построение матрицы ошибок и тепловой карты для бустинг-модели

    Параметры:
    model - обученная бустинг-модель
    test_df - тестовый датафрейм
    label_encoder - LabelEncoder, использованный для кодирования меток
    feature_names - список имен признаков, использованных при обучении
    """
    try:
        # 1. Получаем предсказания
        X_test = test_df[feature_names]
        y_true = test_df['Label_encoded']
        y_pred = model.predict(X_test)

        # 2. Матрица ошибок (абсолютные значения)
        classes = label_encoder.classes_
        cm = confusion_matrix(y_true, y_pred)

        # 3. Нормализованная матрица ошибок (по строкам)
        cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

        # 4. Создаем DataFrame для удобного отображения
        cm_df = pd.DataFrame(cm_normalized,
                            index=classes,
                            columns=classes)

        # 5. Визуализация - тепловая карта
        plt.figure(figsize=(14, 12))
        ax = sns.heatmap(cm_df,
                        annot=True,
                        fmt=".2f",
                        cmap="YlOrRd",
                        cbar_kws={'label': 'Доля предсказаний'},
                        linewidths=0.5,
                        annot_kws={"size": 12},
                        vmin=0, vmax=1)

        # 6. Настройка отображения
        plt.title("Нормализованная матрица ошибок (XGBoost)\n", pad=25, fontsize=16)
        plt.xlabel("Предсказанные классы", fontsize=14)
        plt.ylabel("Истинные классы", fontsize=14)
        ax.xaxis.tick_top()
        ax.xaxis.set_label_position('top')
        plt.xticks(rotation=45, ha='left')

        plt.tight_layout()
        plt.savefig('confusion_matrix_xgboost.png', dpi=150, bbox_inches='tight')
        plt.show()


    except Exception as e:
        print(f"Ошибка при построении матрицы: {str(e)}")
        import traceback
        traceback.print_exc()

# Модифицируем функцию boosting_pipeline для возврата feature_names
def boosting_pipeline(train_df, test_df, model_type='xgb'):
    train_df, test_df, le = prepare_data(train_df, test_df)
    X_train, y_train = balance_data(train_df, le)
    model, feature_names = train_boosting(X_train, y_train, model_type)

    X_test = test_df[feature_names]
    y_test = test_df['Label_encoded']
    y_pred = model.predict(X_test)


    # Вызываем функцию визуализации
    plot_confusion_matrix_and_heatmap(model, test_df, le, feature_names)

    return model, le

# Запуск пайплайна
model, label_encoder = boosting_pipeline(train_df.copy(), test_df.copy(), model_type='xgb')
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE
from autogluon.tabular import TabularPredictor
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

def prepare_data(train_df, test_df):
    """Подготовка данных"""
    # Удаляем лишние колонки
    for df in [train_df, test_df]:
        cols_to_drop = [col for col in ['is_attack', 'Unnamed: 0'] if col in df.columns]
        if cols_to_drop:
            df.drop(cols_to_drop, axis=1, inplace=True)

    # Объединяем редкие классы
    rare_classes = ['NetBIOS', 'LDAP']
    train_df['Label'] = train_df['Label'].replace({k: 'RARE_ATTACK' for k in rare_classes})
    test_df['Label'] = test_df['Label'].replace({k: 'RARE_ATTACK' for k in rare_classes})

    # Фиксируем кодировщик
    le = LabelEncoder()
    le.fit(train_df['Label'])

    # Кодируем метки
    train_df['Label_encoded'] = le.transform(train_df['Label'])
    test_df['Label_encoded'] = le.transform(test_df['Label'])

    return train_df, test_df, le

def balance_data(train_df, le, min_samples=100):
    """SMOTE балансировка с проверкой минимального количества"""
    X = train_df.drop(['Label', 'Label_encoded'], axis=1)
    y = train_df['Label_encoded']

    # Стратегия увеличения для миноритарных классов
    strategy = {}
    for class_id in np.unique(y):
        count = sum(y == class_id)
        if min_samples <= count < 30000:  # Балансируем только средние классы
            strategy[class_id] = min(count * 2, 50000)  # Максимум x2 увеличение

    if strategy:
        smote = SMOTE(sampling_strategy=strategy, k_neighbors=3, random_state=42)
        X, y = smote.fit_resample(X, y)

    balanced_data = pd.DataFrame(X)
    balanced_data['Label_encoded'] = y
    return balanced_data

def plot_confusion_matrix(y_true, y_pred, label_encoder, title_suffix=''):
    """Визуализация нормализованной матрицы ошибок"""
    cm = confusion_matrix(y_true, y_pred)
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]  # Нормализация по строкам (recall)

    plt.figure(figsize=(12, 10))
    sns.set(font_scale=1.1)
    ax = sns.heatmap(
        cm_normalized,
        annot=True,
        fmt='.2f',
        cmap='coolwarm',
        cbar=True,
        vmin=0,
        vmax=1,
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_
    )

    plt.title(f'Нормализованная матрица ошибок (Recall){title_suffix}', fontsize=14)
    plt.xlabel('Предсказанные метки', fontsize=12)
    plt.ylabel('Истинные метки', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()

def train_autogluon(train_data, time_limit=3660):
    """Настройка и обучение AutoGluon с оптимизацией памяти"""
    predictor = TabularPredictor(
        label='Label_encoded',
        problem_type='multiclass',
        eval_metric='f1_weighted',
        verbosity=2
    ).fit(
        train_data,
        presets='medium_quality_faster_train',
        time_limit=time_limit,
        hyperparameters={
            'GBM': [
                {'extra_trees': True, 'ag_args': {'name_suffix': 'XT'}},
                {},
            ],
            'CAT': {'iterations': 1000, 'learning_rate': 0.03},
            'XGB': {'tree_method': 'gpu_hist'},  # Используем GPU если доступно
            'RF': {'max_depth': 15, 'n_estimators': 200}
        },
        excluded_model_types=['KNN', 'NN'],  # Исключаем медленные модели
        save_space=True  # Оптимизация использования диска
    )
    return predictor

def evaluate_model(predictor, test_df, label_encoder):
    """Расширенная оценка модели"""
    # Получаем тестовые данные
    X_test = test_df.drop(['Label', 'Label_encoded'], axis=1, errors='ignore')
    y_test = test_df['Label_encoded']

    # Предсказание с вероятностями
    y_pred = predictor.predict(X_test)
    y_proba = predictor.predict_proba(X_test)

    # Подробный отчет
    print("\n" + "="*60)
    print("Полная оценка модели")
    print("="*60)
    print(predictor.leaderboard(silent=True))

    # Classification report
    print("\nОтчет о классификации:")
    print(classification_report(
        y_test,
        y_pred,
        target_names=label_encoder.classes_,
        digits=4
    ))

    # Визуализация матрицы ошибок
    plot_confusion_matrix(y_test, y_pred, label_encoder)

    return y_test, y_pred

def autogluon_pipeline(train_df, test_df, time_limit=60):
    """Полный пайплайн"""
    # 1. Подготовка данных
    print("[1/4] Подготовка данных...")
    train_df, test_df, le = prepare_data(train_df, test_df)

    # 2. Балансировка классов
    print("\n[2/4] Балансировка данных...")
    balanced_train = balance_data(train_df, le)

    # 3. Обучение модели
    print("\n[3/4] Обучение AutoGluon...")
    predictor = train_autogluon(balanced_train, time_limit)

    # 4. Оценка
    print("\n[4/4] Оценка модели...")
    y_test, y_pred = evaluate_model(predictor, test_df, le)

    return predictor, le, y_test, y_pred

# Запуск пайплайна
if __name__ == '__main__':

    # Выполнение пайплайна
    predictor, label_encoder, y_test, y_pred = autogluon_pipeline(
        train_df.copy(),
        test_df.copy(),
        time_limit=3660  # 1 час на обучение
    )

    # Дополнительная информация
    print("\nСтатистика тестового набора:")
    print(pd.Series(label_encoder.inverse_transform(y_test)).value_counts())
```
